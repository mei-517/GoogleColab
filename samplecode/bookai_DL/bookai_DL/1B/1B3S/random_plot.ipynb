```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import svm
from sklearn.svm import LinearSVC
from sklearn.datasets import make_blobs, make_moons, make_circles
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
#X, y = make_blobs(n_samples=40, centers=[(0,0),(0,0)], n_features=2, cluster_std=1.0, random_state=5)
#X, y = make_blobs(n_samples=40, centers=[(0,0),(1,1)], n_features=2, cluster_std=0.5, random_state=1)
#Fig a
#X, y = make_moons(n_samples=40, noise=0.2, random_state=1)
#Fig b
#X, y = make_circles(n_samples=40, noise=0.1, factor=0.5, random_state=1)
#X, y = make_moons(n_samples=100, noise=0.2, random_state=5)
#X, y = make_blobs(n_samples=40, centers=[(0,0),(1,1),(0,1),(1,0)], n_features=4, cluster_std=0.5, random_state=1)
#Fig c
#X1, y1 = make_blobs(n_samples=40, centers=[(0,0),(0,1.5)], n_features=2, cluster_std=0.3, random_state=2)
#X2, y2 = make_circles(n_samples=40, noise=0.1, factor=0.5, random_state=3)
#X2 = X2+[1.5,0.5]
#y2 = y2 + 2
#X = np.append(X1,X2, axis=0)
#y = np.append(y1,y2, axis=0)
np.random.seed(0)
n = 50
X1 = np.random.uniform(low=-1, high=1, size=(n, 2))
y1 = np.zeros(n)
X2 = np.random.uniform(low=-1, high=1, size=(n, 2))
y2 = np.zeros(n)+1
Xd = np.append(X1,X2, axis=0)
yd = np.append(y1,y2, axis=0)
```

```python
X = Xd
y = yd
clf = svm.SVC(C=100.0, gamma=20.0)
result = clf.fit(X, y)
y_pred = clf.predict(X)
accuracy_score(y_pred,y)
```

```python
fig = plt.figure(1, figsize=(5, 5))
plt.clf()

markers = ['o', 'x', ',', '^', 'v', '*', '+', 'D']
colors =  ['r', 'b', 'g', 'k', 'c', 'm', 'y', 'grey']
m = 0
for i in set(y):
    n = np.where(y==i)
    plt.scatter(X[n, 0], X[n, 1], marker=markers[m], c=colors[m], zorder=10, edgecolors='k')
    m=m+1

plt.axis('tight')
x_min = min(X[:,0])
x_max = max(X[:,0])
y_min = min(X[:,1])
y_max = max(X[:,1])

XX, YY = np.mgrid[x_min:x_max:200j, y_min:y_max:200j]
#Z = clf.decision_function(np.c_[XX.ravel(), YY.ravel()])
Z = clf.predict(np.c_[XX.ravel(), YY.ravel()])

# Put the result into a color plot
Z = Z.reshape(XX.shape)

plt.pcolormesh(XX, YY, Z, cmap=plt.cm.Pastel1)
plt.contour(XX, YY, Z, colors='k')

plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)

plt.xticks(())
plt.yticks(())

plt.show()
fig.savefig("img.png")
fig.savefig("img.pdf")
```

```python
X_train, X_test, y_train, y_test = train_test_split(Xd, yd, test_size=0.2, random_state=5)
```

```python
X = X_train
y = y_train
clf = svm.SVC(C=100.0, gamma=20.0)
result = clf.fit(X, y)
y_pred = clf.predict(X_train)
print(accuracy_score(y_pred,y_train))
y_pred = clf.predict(X_test)
print(accuracy_score(y_pred,y_test))
```
