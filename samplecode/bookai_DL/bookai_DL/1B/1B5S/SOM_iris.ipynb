# ライブラリのインポート

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import somoclu
import pandas as pd
%matplotlib inline
```

# 学習データの読み込み

```python
from sklearn.datasets import load_iris
iris = load_iris()
X = iris.data
Y = iris.target
```

# SOMの設定

```python
n_rows, n_columns = 50, 50
codebook = np.zeros((n_rows*n_columns, X.shape[1]), dtype=np.float32)
som = somoclu.Somoclu(n_columns, n_rows, initialcodebook=codebook)
%time som.train(X)
```

# SOMの学習後の分類結果

```python
som.view_umatrix(figsize =(5,5),bestmatches=True, labels=Y,colormap ='Greys')
```

# クラスタ分析の準備

```python
from sklearn.cluster import KMeans
som.cluster(KMeans(n_clusters=3))
```

# 学習データのクラスの設定

```python
som.clusters
```

# 学習データのクラスの設定

```python
colors =  ['r', 'b', 'g', 'k', 'c', 'm', 'y', 'grey']
a = list(map(lambda x:colors[som.clusters[som.bmus[x][1],som.bmus[x][0]]],range(len(som.bmus))))
print(a)
```

# 学習データのクラス分類結果の表示

```python
som.view_umatrix(figsize =(5,5),bestmatches=True, labels=Y,colormap ='Greys', bestmatchcolors=a)
```
