# 画像変換用プログラム
2つの変換があるので対象とする方を使用してください

## ライブラリのインポート

```python
import numpy as np
import cv2 
import glob
from PIL import Image
```

## PNGファイルからJPEGファイルへの変換

```python
image_list = glob.glob('angle64x64/*/*/*.png')#pngファイルの読み込み
for f in image_list:
    img = Image.open(f)
    r, g, b, a = img.split()
    a = np.array(a)
    a = cv2.resize(a, dsize=(32, 32))#サイズを32×32ピクセルに
    fc = f.replace('png', 'jpg')#jpgファイルへの変換
    cv2.imwrite(fc, a)
    print(fc)
```

## 画像ファイルのサイズ変更

```python
image_list = glob.glob('angle64x64/*/*/*.jpg')#この例ではjpgファイルを対象としている
for f in image_list:
    img = Image.open(f)
    img = img.resize((32, 32))#サイズを32×32ピクセルに
    img.save(f)
    print(f)
```
