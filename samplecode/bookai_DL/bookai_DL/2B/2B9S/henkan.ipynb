#訓練データと単語リスト作成用プログラム

## 作品名の設定

```python
book_name = 'momotaro'
```

## 文書の読み込み

```python
with open(book_name+'_remove.txt', encoding='utf-8') as f:
    story = f.read()
#print(story)
```

## 不要な部分の削除用ライブラリのインポート

```python
import re
```

##不要な部分の削除

・全角スペース（各文の最初にあります。）

・《　》のカッコとその中にある文字（ルビを表しています。）

・［　］のカッコとその中にある文字（文章の体裁に関して書かれています。）


```python
story = re.sub("　", "", story)#全角スペース
story = re.sub("\《.+?\》", "", story)
story = re.sub("\［.+?\］", "", story)

print(story)
```

##分かち書き用ライブラリのインストール

```python
import json
import MeCab
```

##分かち書き

```python
tagger = MeCab.Tagger('-Owakati')
#tagger = MeCab.Tagger()
story_list = tagger.parse(story)
print(story_list)
```

## 訓練データの保存

```python
with open(book_name+'_train.txt', mode='w') as f:
  f.write(story_list)
```

##単語の重複をまとめて、出現回数の多い順に並べる

```python
from collections import Counter
story_word = Counter(story_list.split()).most_common()
print(story_word)
```

## 単語データの保存

```python
with open(book_name+'_vocab.txt', mode='w') as f:
  for sw in story_word:
    f.write(sw[0]+'\n')
```
