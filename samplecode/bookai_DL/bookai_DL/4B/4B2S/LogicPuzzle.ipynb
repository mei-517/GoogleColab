## ルートを表示するための関数

```python
def ShowPuzzle(pzl):
    for i in range(5):
        print(start_ball[i], end=' ')
    print('')
    for j in range(6):
        for i in range(5):
            print(pzl[j][i], end=' ')
        print('')
    for i in range(5):
        print(goal_ball[i], end=' ')
    print('')
```

## 正しいルートとなっているかを調べる関数

すべての交差ピースを使っていてかつ正しいルートだった場合はTrueを戻り値とする。
これまで見つかったすべての交差ピースを使ったパターンの数（tolal_num）と、使用している交差ピースの数（cb_num）も戻り値としている。

```python
def Check(pzl, tolal_num):
    cb_num = 0
    for i in range(6):
        cb_num = cb_num + pzl[i].count(2)
    if cb_num<cross_block:
        return False, tolal_num, cb_num
    tolal_num = tolal_num + 1
    temp_ball = copy.copy(start_ball)
    for j in range(6):
        for i in range(5):
            if pzl[j][i] == 1 or pzl[j][i] == 0: temp_ball[i] = temp_ball[i]
            elif pzl[j][i] == 2: temp_ball[i], temp_ball[i+1] = temp_ball[i+1], temp_ball[i]
    return (temp_ball == goal_ball), tolal_num, cb_num
```

## 探索

正しいルートとなるパターンが見つかるたびに表示する。
最後には、すべてのピースを使って作成できるパターンの合計数が表示される。

```python
open_list = [init_pzl]
closed_list = []

total_num = 0
while open_list:
    print('\r'+str(total_num)+'\t'+str(len(open_list))+'\t'+str(len(closed_list))+'\t',end='')
    ol = open_list.pop(-1)#深さ優先
    closed_list.append(ol)
    f,total_num,cb_num= Check(ol, total_num)

    if f:
        print('\n'+'goal')
        ShowPuzzle(ol)
        
    if cb_num < cross_block:
        for i in range(6):
            if ol[i].count(0)>1:
                for j in range(4):
                    if ol[i][j] == 0 and ol[i][j+1] == 0:
                        new_pzl = copy.deepcopy(ol)
                        new_pzl[i][j] = 2 
                        new_pzl[i][j+1] = 3 
                        if not(new_pzl in closed_list) and not(new_pzl in open_list) :
                            open_list.append(new_pzl)
print('\n'+str(total_num))
```
