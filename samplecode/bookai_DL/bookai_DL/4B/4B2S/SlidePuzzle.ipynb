# スライドパズル

## ライブラリのインポート

```python
import numpy as np
import copy
```

## パズルのサイズの設定

```python
size = 3
```

## 各マスで動ける方向の設定

```python
move = []
for n in range(size**2):
    tmp = []
    if n%size!=0:
        tmp.append(n-1)
    if (n+1)%size!=0:
        tmp.append(n+1)
    if n>size-1:
        tmp.append(n-size)
    if n<size*size-size:
        tmp.append(n+size)
#    print(n, tmp)
    move.append(tmp)
print(move)        
```

## パズルを表示するための関数

```python
def ShowPuzzle(p):
    for i in range(size**2):
        if i%size==size-1:
            print(p[i])
        else:
            print(p[i], end=' ')
```

## 初期配置

```python
goal_pzl = list(range(1,size**2))
goal_pzl.append(0)
import random
random.seed(1)

p = copy.copy(goal_pzl)
for i in range(10):
    now = p.index(0)
    j = random.choice(move[now])
    p[now],p[j] = p[j],p[now]
init_pzl = p

ShowPuzzle(init_pzl)
```

## 探索

```python
init_dict = {'id':0,'pzl':init_pzl,'prt':-1,'chd':[],'depth':0}

open_list = [init_dict]
search_list = [init_dict]
pzl_list = [init_pzl]

while open_list:
    if goal_pzl in pzl_list:
        break
    ol = open_list.pop(0)#幅優先
#    ol = open_list.pop(-1)#深さ優先
    now = ol['pzl'].index(0)#0の位置をnowに入れる
    print("\r"+str(ol['depth'])+"\t"+str(ol['id']),end="")
    for i in move[now]:
        p = copy.copy(ol['pzl'])
        p[now],p[i] = p[i],p[now]
        if not(p in pzl_list):
            #新しい盤面だった場合の処理
            pzl_list.append(p)
            pzl_id = len(pzl_list)-1
            pzl_pzl = p
            pzl_prt = ol['id']
            pzl_depth = ol['depth']+1
            ol['chd'].append(pzl_id)
            pzl_dict = {'id':pzl_id,'pzl':pzl_pzl,'prt':pzl_prt,'chd':[],'depth':pzl_depth}
            open_list.append(pzl_dict)
            search_list.append(pzl_dict)
            if goal_pzl in pzl_list:
                goal_id = pzl_id
                break
            

print('\ngoal_id',goal_id)

n = goal_id
while n!=-1:
    for sl in search_list:
        if sl['id'] == n:
            print('depth',sl['depth'],'id',sl['id'])
            ShowPuzzle(sl['pzl'])
            n = sl['prt']
            break
```
