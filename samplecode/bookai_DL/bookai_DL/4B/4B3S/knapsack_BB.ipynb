# ナップザック問題

## ライブラリのインポート

```python
import numpy as np
import copy
```

## 最大の重さと商品の設定

```python
MAX_WEIGHT = 10#12#
items=[(1, 20), (2, 30), (4, 50), (5, 40), (7, 60)]
#items=[(1, 20), (2, 30), (4, 50), (5, 40), (7, 60), (3, 20), (6, 100)]
```

## 比率順に並べ替え

```python
items = [[items[i][0],items[i][1],items[i][1]/items[i][0]] for i in range(len(items))]
items = sorted(items, reverse=True, key=lambda x: x[2])  #[1]に注目してソート
print(items)
```

## データから重さと価値を計算

重さがオーバーした場合は価値を-10にする

```python
def Check(bag):
    w,v = 0, 0
    for i in range(len(bag)):
        if bag[i]==1:
            w += items[i][0]
            v += items[i][1]
    if w>MAX_WEIGHT:
        v = -v
    return w, v
```

## データから最小の価値と最大の価値を求める

```python
def ValueMinMax(bag):
    tmp_bag =  copy.copy(bag)
    v_min = 0
    v_max = 0
    for i in range(len(items)):
        if tmp_bag[i]==-1:
            tmp_bag[i] = 1
        w,v = Check(tmp_bag)
        if v<0 or (w==MAX_WEIGHT and tmp_bag.count(-1)==0):
            if v<0:
                tmp_bag[i] = 0
            w_min,v_min = Check(tmp_bag)
            v_max = v_min+(float(MAX_WEIGHT-w_min)/items[i][0]*items[i][1])
            return tmp_bag,i, w_min, v_min, v_max
    print(w)
```

## 探索

```python
init_bag = [-1]*len(items)
max_bag = init_bag
max_weight = 0
max_value = 0

open_list = [init_bag]
closed_list = []

while open_list:
    bag = open_list.pop(0)#幅優先
#    bag = open_list.pop(-1)#深さ優先
    closed_list.append(bag)
    print("bag",bag,len(open_list))

    a = ValueMinMax(bag)
    print(a)
    if a!=None:
        now_bag,n,w,v_min,v_max = a[0],a[1],a[2],a[3],a[4]
        if v_min==v_max:
            if max_value < v_min:
                max_bag = now_bag
                max_weight = w
                max_value = v_min
                print("max",max_bag, max_weight, max_value)
        else:
            for i in range(2):
                new_bag = copy.copy(bag)
                new_bag[n] = i
                if not(new_bag in open_list) and not(new_bag in closed_list):
                    open_list.append(new_bag)
                    print("add",new_bag)

print('探索数')
print(len(closed_list))
print('答え')
print('商品の並び順')
print(items)
print('選択された商品')
print(max_bag)
print('合計の重さ')
print(max_weight)
print('最大の価値')
print(max_value)
```
